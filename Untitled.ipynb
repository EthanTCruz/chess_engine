```python
import sys
sys.path.append("../")
from torch.utils.data import Dataset, DataLoader
from sqlalchemy.orm import sessionmaker
from sqlalchemy import create_engine
import torch
import numpy as np
from chess_engine.src.model.classes.bitboard_processing.bitboard_creator import bitboards_to_array, sample_bitboard_dict, Bitboard_Creator
from chess_engine.src.model.classes.sqlite.database import  get_db
from chess_engine.src.model.classes.sqlite.models import (GamePositions,
                                                          GamePositionRollup,
                                                          TrainGamePositions,
                                                          ValidationGamePositions,
                                                          TestGamePositions)
```

```python
model = TrainGamePositions
```

```python
with next(get_db()) as session:
    length = session.query(model).count()
    
```

```python
idx = 0
batch_size = 128
bitboards = None
label = None
with next(get_db()) as session:
    while True:
        # Fetch a batch of records
        records = session.query(model).offset(idx).limit(batch_size).all()
        
        if not records:  # If no records are found, exit the loop
            print(f"No more records found starting at index {idx}")
            break
        
        for record in records:
            # Extract attributes based on the provided dictionary
            features = [getattr(record, attr) for attr in sample_bitboard_dict.keys()]
            
            # Generate bitboards for the record
            bitboards = bitboards_to_array(features)
            

            
            # Set up labels if necessary
            label = np.array(record.win_buckets)
            
            # Process the tensor and label (example)
            # print(f"Processed record at index {idx}: {bitboard_tensor}, label: {label}")
            break
        break    
        # Increment idx by the batch size
        idx += batch_size
```
