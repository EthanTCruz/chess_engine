```python
import torch
import sys
sys.path.append("../")
from chess_engine.src.model.classes.basic_model import ChessEvalCNN
from chess_engine.src.model.classes.bitboard_processing.bitboard_creator import sample_bitboard_dict
from chess_engine.src.model.config.config import  model_settings, data_settings
from chess_engine.src.model.classes.bitboard_processing.bitboard_creator import sample_bitboard_dict
from chess_engine.src.model.classes.h5py_piping.dataloader import get_dataloader, HDF5SingleFileDataset
from chess_engine.src.model.classes.basic_model import ChessEvalCNN
from chess_engine.src.model.classes.bitboard_processing.bitboard_creator import sample_bitboard_dict
import random
import numpy as np

```

```python
# Define the model path (ensure this matches your saved model path)
model_path = model_settings.torch_model_file

# Load the model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ChessEvalCNN(in_channels=len(sample_bitboard_dict)).to(device)

# Load the checkpoint
checkpoint = torch.load(model_path, map_location=device)
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()  # Set the model to evaluation mode

print(f"Model loaded from {model_path}")
```

```python
# Example usage of the model
def prepare_input(bitboard):
    """
    Prepare the input tensor for the model from the given bitboard.
    """
    tensor_input = torch.tensor(bitboard, dtype=torch.float32).unsqueeze(0).to(device)
    return tensor_input

# Assume you have a bitboard dictionary for a chess position
bitboard = sample_bitboard_dict  # Replace with your actual bitboard data
input_tensor = prepare_input(bitboard)

# Run the model
with torch.no_grad():
    output = model(input_tensor)

print(f"Model output: {output}")
```
