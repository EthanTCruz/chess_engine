```python
def en_passant_bitboard(fen: str) -> int:
    """
    Given a FEN string, returns the bitboard representation of the en passant square.
    If there is no en passant square, returns 0.
    """
    board = chess.Board(fen)
    ep_square = board.ep_square  # Get the en passant square index
    
    if ep_square is None:
        return 0
    
    return 1 << ep_square  # Convert to bitboard format

def bitboard_to_string(bitboard: int) -> str:
    """
    Convert a bitboard integer into a human-readable 8x8 board representation.
    """
    board_str = ""
    for rank in range(7, -1, -1):
        for file in range(8):
            square = rank * 8 + file
            if (bitboard >> square) & 1:
                board_str += "1 "  # Mark en passant square
            else:
                board_str += ". "  # Empty square
        board_str += "\n"
    return board_str
    
```
